```python
import yaml
from ultralytics import YOLO


yaml_data = {
    'train': [
        '/kaggle/input/datasets/hiruthiksudhakar/fabric-dataset/train/images',  
        '/kaggle/input/datasets/hiruthiksudhakar/fabric-dataset-2/lim.v3i.yolo26/train/images'      
    ],
    'val': [
        '/kaggle/input/datasets/hiruthiksudhakar/fabric-dataset/valid/images',
        '/kaggle/input/datasets/hiruthiksudhakar/fabric-dataset-2/lim.v3i.yolo26/valid/images'
    ],
    'nc': 4,
    'names': ['Hole', 'Knot', 'Line', 'Stain']
}

ghost_yaml_path = '/kaggle/working/ghost_data_merged.yaml'

with open(ghost_yaml_path, 'w') as f:
    yaml.dump(yaml_data, f, default_flow_style=False)

print("✅ Multi-Dataset Ghost YAML created!")

```

```python
model = YOLO('/kaggle/input/models/hiruthiksudhakar/fabric-model-1/pytorch/default/1/best.pt') 
```

```python
results = model.train(
    data=ghost_yaml_path,  
    epochs=100,      
    imgsz=640,
    batch=32,        
    device=[0, 1],  
    augment=True,    
    degrees=15.0,    
    fliplr=0.5,      
    flipud=0.5,      
    hsv_s=0.4,       
    hsv_v=0.4,       
    project="/kaggle/working/TextileIQ_Phase2", 
    name="yolo26m_augmented"
)
```

```python
model.export(format="onnx")
```
